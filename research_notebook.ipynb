```python
import pandas as pd
symbol = "EURUSD"
tf = mt5.TIMEFRAME_H1
from_position = 1
num_bars= 1 #ultima vela cerrada

bars = pd.DataFrame(mt5.copy_rates_from_pos(symbol, tf, from_position, num_bars)) # Copiar los datos de las velas
bars['time'] = pd.to_datetime(bars['time'], unit='s') #formato del tiempo
bars.set_index('time', inplace=True) #establecer el tiempo como indice

bars.rename(columns={'tick_volume': 'tickvol', 'real_volume': 'vol'}, inplace=True)
bars = bars[['open', 'high', 'low', 'close', 'tickvol', 'vol', 'spread']]
bars
```

```python
def get_latest_tick(symbol: str) -> dict:
        try:
            tick = mt5.symbol_info_tick(symbol)
            if tick is None:
                print(f" No se ha podido recuperar el último tick de {symbol} - MT5 error: {mt5.last_error()}")
                return {} #devuelve un diccionario vacío si no se puede recuperar el tick
        
        except Exception as e:
            print(f"Algo no ha ido bien a la hora de recuperar el último tick de {symbol}. MT5 error: {mt5.last_error()}, exception: {e}")
        
        else:
            return tick._asdict()

get_latest_tick("EURUSD")
```
